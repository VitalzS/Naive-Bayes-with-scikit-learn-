`CountVectorizer` turns each document into word counts. `MultinomialNB` uses those counts to learn a classifier. The output shows the bag-of-words tables, the predicted label, and each class's predicted probability. No hand calculations are required.


````python
# Import pandas to show the word-count vectors in readable tables.
import pandas as pd

# Import CountVectorizer to turn text into a bag of word counts.
from sklearn.feature_extraction.text import CountVectorizer

# Import the multinomial Naive Bayes classifier for word-count features.
from sklearn.naive_bayes import MultinomialNB
````

````python
# Enter each training sentence together with its correct class label.
###training_data = [
###    ("Chinese Beijing Chinese", "China"),
###    ("Chinese Chinese Shanghai", "China"),
###    ("Chinese Macao", "China"),
###    ("Tokyo Japan Chinese", "Japan"),
###]

# Enter the new sentence whose class you want to predict.
### test_document = "Chinese Chinese Chinese Tokyo Japan"

training_data = [
("The team won the soccer game", "Sports"),
("The stadium hosted a soccer match", "Sports"),
("Heavy rain delayed the outdoor event", "Weather"),
("The forecast predicts rain tomorrow", "Weather"),
]


test_document = "Rain rain delayed the soccer game"
````

````python
# Separate the training sentences from their labels for scikit-learn.
documents = [sentence for sentence, label in training_data]
labels = [label for sentence, label in training_data]

# Set up word counting: lowercase text and count words of one or more letters.
vectorizer = CountVectorizer(lowercase=True, token_pattern=r"(?u)\b[a-zA-Z]+\b")

# Learn the vocabulary from training sentences and count their word occurrences.
X_train = vectorizer.fit_transform(documents)

# Count words in the test sentence using only the training vocabulary.
X_test = vectorizer.transform([test_document])

# Get the vocabulary words in the same order as the matrix columns.
words = vectorizer.get_feature_names_out()
````

````python
# Show the training documents as a bag-of-words table.
print("Training documents (word counts):")
display(pd.DataFrame(X_train.toarray(), columns=words, index=labels))
````

````python
# Show the test sentence in the same bag-of-words columns.
print("Test document (word counts):")
display(pd.DataFrame(X_test.toarray(), columns=words, index=["test"]))
````

````python

# Create a Naive Bayes model with add-one smoothing (alpha = 1).
model = MultinomialNB(alpha=1.0)

# Learn class priors and word probabilities from the labeled training counts.
model.fit(X_train, labels)

# Predict the most likely class for the test sentence.
prediction = model.predict(X_test)[0]

# Calculate the normalized predicted probability for every class.
probabilities = model.predict_proba(X_test)[0]

# Show the original sentence and the predicted class.
print("Test sentence:", test_document)
print("Predicted class:", prediction)

# Show class labels and probabilities side by side.
print("Predicted probabilities:")
display(pd.DataFrame({"class": model.classes_, "probability": probabilities}))
````
